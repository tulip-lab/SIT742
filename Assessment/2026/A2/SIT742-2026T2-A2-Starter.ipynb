General lag rule: for each forecast month `m`, the naive source month is `m - lag` months. If the source month is at or before `cutoff_label`, use the observed actual value. `lag=1` is a previous-month naive forecast and normally becomes recursive carry-forward under a fixed cutoff. `lag=12` is a same-month-previous-year naive forecast when the required source months are available before the cutoff.
