#### 4.1 Lists

A list is a sequence of values. The values in a list can be any type. We call the values `items` or `elements`.


````python
shoplist = ['apple', 'mango', 'carrot', 'banana']
l = [10, 20, 30, 40]
empty = []  # Initialize an empty list
````

````python
shoplist
````

````python
l
````

````python
empty
````

````python
shoplist
````

Lists can hold different types, including another list. Use square brackets and an index to access an item. Negative indexes count from the right.


````python
l = [10, 20, 30, 40]
````

````python
l[2]
````

````python
l[-1]
````

````python
l[-3]
````

A nested list is a list stored inside another list. Use one index to access the nested list and another index to access an item inside it.


````python
l = ['apple', 2.0, 5, [10, 20]]
````

````python
l[1]
````

````python
l[3]
````

````python
l[3][1]
````

Unlike strings, lists are mutable. You can assign a new value to a list item.


````python
l = [10, 20, 30, 40]
l[1] = 200
````

````python
l
````

#### 4.2 List operations and slices

Use `in` to test membership. The result is `True` or `False`.


````python
shoplist = ['apple', 'mango', 'carrot', 'banana']
````

````python
'apple' in shoplist
````

````python
'rice' in shoplist
````

The `in` operator also applies to strings.


````python
'a' in 'banana'
````

````python
'seed' in 'banana'  # Test if 'seed' is a substring of 'banana'
````

The `+` operator concatenates lists. The `*` operator repeats a list a given number of times.


````python
[10, 20, 30, 40] + [50, 60]
````

Original expression for list repetition:

```python
[50, 60] * 3
```

Run it in the code cell below to inspect the result.


````python
[50, 60] * 3
````

Slicing retrieves part of a list. The first index is included and the second index is excluded.


````python
l = [1, 2, 3, 4]
````

````python
l[1:3]  # From position 1 to position 3 (excluded)
````

````python
l[:2]  # From the beginning to position 2 (excluded)
````

````python
l[-2:]  # From the second right to the end of the list
````

If you omit both indices, the slice is a copy of the whole list.


````python
l[:]
````

Copying can be useful before modifying the original list.


````python
l = [1, 2, 3, 4]
l_org = l[:]
l[0] = 8
````

````python
l
````

````python
l_org  # the original list is unchanged
````

#### 4.3 List methods and traversal

Common list-related methods and functions include `append()`, `len()`, `sort()`, `split()`, and `join()`.


`append()` adds a new element to the end of a list.


````python
l = [1, 2, 3, 4]
````

````python
l
````

````python
l.append(5)
l.append([6, 7])  # list [6, 7] is nested in list l
````

````python
l
````

`len()` returns the number of items in a list. A nested list counts as one item.


````python
l = [1, 2, 3, 4, 5]
len(l)
````

````python
# A list nested in another list is counted as a single item
l = [1, 2, 3, 4, 5, [6, 7]]
len(l)
````

`sort()` modifies a list in place and returns `None`. `sorted()` returns a new sorted list and leaves the original list unchanged.


````python
shoplist = ['apple', 'mango', 'carrot', 'banana']
shoplist.sort()
````

````python
shoplist
````

````python
shoplist = ['apple', 'mango', 'carrot', 'banana']
shoplist_sorted = shoplist.sort()
````

````python
shoplist_sorted  # No value is returned
````

````python
shoplist = ['apple', 'mango', 'carrot', 'banana']
shoplist_sorted = sorted(shoplist)  # sorted() function returns a new list
````

````python
shoplist_sorted
````

````python
shoplist
````

`split()` converts a string to a list of pieces. `join()` converts a list of strings back to one string.


````python
s = 'I love apples'
s.split(' ')
````

````python
s = 'spam-spam-spam'  # A delimiter '-' is specified here. It is used as word boundary
s.split('-')
````

````python
l = ['I', 'love', 'apples']
s = ' '.join(l)
s
````

A `for` loop is the most common way to traverse a list. Use indexes when you need to update items in place.


````python
shoplist = ['apple', 'mango', 'carrot', 'banana']
for item in shoplist:
    print(item)
````

````python
l = [2, 3, 5, 7]
for i in range(len(l)):
    l[i] = l[i] * 2
print(l)
````

#### 4.4 Tuples

A tuple is also a sequence of values. Tuples and lists are similar, but tuples are immutable: after a tuple is created, its items cannot be changed.


````python
year_born = ('David Hilton', 1995)
year_born
````

To define a tuple, use values separated by commas. Parentheses are common but not always required. A one-item tuple needs a trailing comma.


````python
t = ()  # Empty tuple
t
````

````python
t = (1)
type(t)  # Its type is int, since no comma is following
````

````python
t = (1,)  # One item tuple; the item needs to be followed by a comma
type(t)
````

Tuple indexing and slicing work like list indexing and slicing.


````python
t = ('a', 'b', 'c', 'd')
````

````python
t[0]
````

````python
t[1:3]
````

The original example below intentionally tries to modify a tuple. It is guarded so the notebook remains runnable while still showing the expected error.


````python
t = ('a', 'b', 'c', 'd')
try:
    t[1] = 'B'
except TypeError as err:
    print(f'Expected tuple immutability error: {err}')
````

Tuple assignment lets several variables receive values from a tuple in one statement.


````python
t = ('David', '0233', 78)
(name, id, score) = t
````

````python
name
````

````python
id
````

````python
score
````

The number of variables on the left must match the number of values on the right. The original mismatch example is guarded so execution can continue.


````python
try:
    (a, b, c, d) = (1, 2, 3)
except ValueError as err:
    print(f'Expected tuple unpacking error: {err}')
````

A list of tuples is common when each item has a fixed structure.


````python
t = [('David', 90), ('John', 88), ('James', 70)]
for (name, score) in t:
    print(name, score)
````

#### 4.5 Dictionaries

A dictionary associates keys with values. In this example, the student name is the key and the score is the value.


````python
dict = {'David': 70, 'John': 60, 'Mike': 85}
dict['David']
dict['Anne'] = 92  # add a new item in the dictionary
dict
````

Dictionaries preserve insertion order in modern Python. You can traverse keys directly or sort the keys first when you need a predictable display order.


````python
dict = {'David': 70, 'John': 60, 'Amy': 85}
for key in dict:
    print(key, dict[key])
````

````python
dict = {'David': 70, 'John': 60, 'Amy': 85}
sortedKeys = sorted(dict.keys())
for key in sortedKeys:
    print(key, dict[key])
````

<a id="5-testing-and-analysis"></a>

### 5. Testing and Analysis

Use small checks to confirm the behaviour of collection types. These visible checks help you reason about mutability, indexing, and lookup behaviour.


````python
sample_list = [10, 20, 30, 40]
assert sample_list[0] == 10
assert sample_list[-1] == 40
assert sample_list[1:3] == [20, 30]

copy_source = [1, 2, 3]
copy_result = copy_source[:]
copy_source[0] = 99
assert copy_result == [1, 2, 3]

sample_tuple = ('name', 42)
assert sample_tuple[0] == 'name'

sample_scores = {'David': 70, 'Anne': 92}
assert sample_scores['Anne'] == 92
assert sorted(sample_scores.keys()) == ['Anne', 'David']

print('Collection checks passed.')
````

Normal case: indexing, slicing, and dictionary lookup return expected values. Edge case: one-item tuples need a trailing comma. Failure case: tuples cannot be updated in place, and tuple unpacking requires matching lengths.
